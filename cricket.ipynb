```python
df = pd.read_csv("batter_player_stats.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
plt.figure()
sns.histplot(df["total_runs"], kde=True)
plt.title("Distribution of Runs")
plt.show()
```

```python
plt.figure()
sns.boxplot(x=df["total_runs"])
plt.title("Boxplot of Runs")
plt.show()
```

```python
plt.figure()
sns.boxplot(x=df["strike_rate"])
plt.title("Boxplot of Strike Rate")
plt.show()
```

```python
plt.figure()
sns.scatterplot(x="average", y="strike_rate", data=df)
plt.title("Average vs Strike Rate")
plt.show()
```

```python
top_runs = df.sort_values(by="total_runs", ascending=False).head(10)

plt.figure(figsize=(8,6))
sns.barplot(x="total_runs", y="player_name", data=top_runs)
plt.title("Top 10 Run Scorers")
plt.show()
```

```python
top_sr = df.sort_values(by="strike_rate", ascending=False).head(10)

plt.figure(figsize=(8,6))
sns.barplot(x="strike_rate", y="player_name", data=top_sr)
plt.title("Top 10 Strike Rate Players")
plt.show()
```

```python
top10_avg = df.sort_values(by="average", ascending=False).head(10)

top10_avg[["player_name", "average"]]
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x="average", y="player_name", data=top10_avg)

plt.title("Top 10 Players with Highest Batting Average")
plt.xlabel("Batting Average")
plt.ylabel("Player")
plt.show()
```
